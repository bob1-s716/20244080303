## 1. 四个运算符

分别计算 `8 / 3`、`8 // 3`、`8 % 3`、`2 ** 7`，并输出结果与类型。

````python
print("8 / 3  =", 8 / 3,  "| type =", type(8 / 3))
print("8 // 3 =", 8 // 3, "| type =", type(8 // 3))
print("8 % 3  =", 8 % 3,  "| type =", type(8 % 3))
print("2 ** 7 =", 2 ** 7, "| type =", type(2 ** 7))
````

**各运算符含义：**

| 运算符 | 名称 | 含义 | 本例结果 |
| --- | --- | --- | --- |
| `/` | 真除法 | 按数学除法计算，结果一定是浮点数 | `2.6666666666666665`（float） |
| `//` | 整除（向下取整除法） | 相除后向下取整，结果为整数或浮点数 | `2`（int） |
| `%` | 取模 | 求余数，满足 `a == (a // b) * b + a % b` | `2`（int） |
| `**` | 幂运算 | 求 `a` 的 `b` 次方 | `128`（int） |

结论：`/` 的结果类型总是 `float`，而 `//`、`%`、`**` 在整数运算中保持 `int`。

## 3. 类型转换与异常

执行 `int("123")`、`float("3.14")`、`int(3.9)`，并捕获 `int("goat")` 的异常。

````python
print('int("123")  =', int("123"),  "| type =", type(int("123")))
print('float("3.14") =', float("3.14"), "| type =", type(float("3.14")))
print("int(3.9)    =", int(3.9),  "| type =", type(int(3.9)))

try:
    int("goat")
except ValueError as e:
    print("\n捕获异常 -> ValueError:", e)
````

**转换结果解释：**

1. `int("123")` 把**只含整数字符串**按十进制解析为整数 `123`；`float("3.14")` 把小数字符串解析为浮点数 `3.14`。
2. `int(3.9)` 的结果是 `3` 而不是 `4`：`int()` 对浮点数做的是**向零截断**（丢弃小数部分），
   而不是四舍五入。想要四舍五入应使用 `round(3.9)`（得 `4`），想要向下取整可以用 `math.floor`。
3. `int("goat")` 抛出 `ValueError`：字符串 `"goat"` 不符合整数的字面格式，`int()` 无法解析，
   这种“类型正确但取值非法”的错误正属于 `ValueError`（若把不可转换的对象传进去，例如
   `int(None)`，抛出的则是 `TypeError`）。

## 4. 字符串常用操作

对 `" Python Data Analysis "` 完成去除首尾空白、大小写转换、切片、单词拆分，再用 `join()` 连接。

````python
raw = " Python Data Analysis "
stripped = raw.strip()

print("原始字符串  :", repr(raw))
print("strip() 后  :", repr(stripped))
print("upper()     :", stripped.upper())
print("lower()     :", stripped.lower())

print("\n切片 [:6]   :", repr(stripped[:6]))
print("切片 [-8:]  :", repr(stripped[-8:]))
print("切片 [::2]  :", repr(stripped[::2]))
print("切片 [::-1] :", repr(stripped[::-1]))

words = stripped.split()
print("\nsplit()     :", words)
print('join("-")   :', "-".join(words))
print('join("")    :', "".join(words))
````

**说明：**

- `strip()` 去掉首尾空白（不改变中间的单个空格），`upper()/lower()` 返回全大写/全小写的新字符串；
- 切片语法 `s[start:stop:step]`，省略 start/stop 表示从头/到尾，`[::-1]` 即反转字符串；
- `split()` 不传参数时按任意空白分割成单词列表，`join()` 是它的逆操作，用指定分隔符把序列连接成字符串。
  注意 `join()` 必须由**分隔符**调用，写成 `"-".join(words)` 而不是 `words.join("-")`。

## 5. 拼接、重复与字符串的不可变性

比较拼接与重复操作，并验证字符串不支持直接修改单个字符，最后给出构造新字符串的方式。

````python
left, right = "Python", "Data"

print("拼接 left + ' ' + right :", left + " " + right)
print("重复 (left + '-') * 3   :", (left + "-") * 3)
print("重复 '-' * 10           :", "-" * 10)
print("拼接后长度:", len(left + " " + right))

text = "Python"
print("\n原字符串:", text, "| id =", id(text))

try:
    text[0] = "J"          # 字符串是 immutable，元素赋值会失败
except TypeError as e:
    print("捕获异常 -> TypeError:", e)

# 构造新字符串的两种方式
new_by_slice = "J" + text[1:]
new_by_replace = text.replace("P", "J")
print("切片构造新字符串:", new_by_slice)
print("replace 构造新字符串:", new_by_replace)
print("原字符串未被改变:", text)
````

**结论：**字符串是**不可变（immutable）**对象，`text[0] = "J"` 会抛出
`TypeError: 'str' object does not support item assignment`。所有“修改”字符串的方法
（`strip`、`upper`、`replace`、`+`、`*`）都返回**新对象**，原对象保持不变；
要构造新字符串可以用切片拼接或 `replace()`。

# 实验2：函数、条件、循环与递归

## 1. `score_level(score)`

90~100 优秀，80~不足 90 良好，60~不足 80 及格，0~不足 60 不及格；超出 0~100 抛 `ValueError`。

````python
def score_level(score):
    """把 0~100 的成绩转换为等级；超出范围抛出 ValueError。"""
    if not 0 <= score <= 100:
        raise ValueError(f"成绩必须在 0~100 之间，收到 {score}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"
````

````python
# 边界值测试
for s in [100, 95, 90, 89, 80, 79, 60, 59, 0]:
    print(f"{s:>3} 分 -> {score_level(s)}")

print()
for bad in [101, -1]:
    try:
        score_level(bad)
    except ValueError as e:
        print(f"score_level({bad}) -> ValueError: {e}")
````

**思路：**先做范围校验，再用 `if / elif` 从高分到低分依次判断。这样写的好处是每个条件
只需写一个边界（`>= 90`、`>= 80`、`>= 60`），后续分支自动落在更低的区间里，不会出现
区间重叠或遗漏。

## 2. 用 `for` 与 `while` 计算 1 到 100 的和

````python
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for 循环结果  :", total_for)
print("while 循环结果:", total_while)
print("两者相等:", total_for == total_while)
print("公式验证 100*101/2 =", 100 * 101 // 2)
````

**比较：**两种实现结果完全一致（`5050`），与求和公式 `n(n+1)/2` 相同。
`for` 适合已知迭代次数的场景（配合 `range`），`while` 适合终止条件由状态决定的场景，
代价是需要自己维护循环变量并保证它能终止。

## 3. 递归版与迭代版阶乘

约定 `0! = 1`，输入为非负整数；比较 `n = 0, 1, 5, 10` 的结果。

````python
def factorial_recursive(n):
    """递归版 n!；n 为非负整数。"""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    """迭代版 n!；n 为非负整数。"""
    if n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


for n in [0, 1, 5, 10]:
    r, it = factorial_recursive(n), factorial_iterative(n)
    print(f"{n:>2}! = 递归 {r:<8} 迭代 {it:<8} 相等: {r == it}")
print("\n10! =", factorial_recursive(10))
````

**比较：**两者结果一致，递归版更贴近 `n! = n × (n-1)!` 的数学定义，迭代版更省内存。
Python 默认递归深度约 1000，`n` 很大时递归会触发 `RecursionError`，工程上更常用迭代。

## 4. `summary(values)`：以元组返回最小值、最大值和平均值

通过元组解包接收结果。

````python
def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 三元组。"""
    if not values:
        raise ValueError("values 不能为空")
    return (min(values), max(values), sum(values) / len(values))


sample = [12, 4, 19, 7, 15]
lowest, highest, mean = summary(sample)      # 元组解包
print("数据    :", sample)
print("最小值  :", lowest)
print("最大值  :", highest)
print("平均值  :", mean)
print("返回类型:", type(summary(sample)))
````

## 5. `sum_all(*args)`：返回所有参数之和

````python
def sum_all(*args):
    """返回所有位置参数之和；无参数时返回 0。"""
    total = 0
    for value in args:
        total += value
    return total


print("无参数      : sum_all()        =", sum_all())
print("单个参数    : sum_all(5)       =", sum_all(5))
print("多个参数    : sum_all(1,2,3,4,5) =", sum_all(1, 2, 3, 4, 5))

numbers = [10, 20, 30]
print("列表拆包传入: sum_all(*numbers) =", sum_all(*numbers), "（等价于 sum_all(10, 20, 30)）")
````

**说明：**`*args` 把任意数量的位置参数收集成**元组**，因此无参数时循环一次也不执行、
返回 0；调用时用 `*numbers` 又能把列表拆成多个位置参数传进去。

## 5. 赋值与 `copy()` 的区别

用**单独的演示列表**做实验，避免改变上面的原始成绩：
`alias = demo` 是取别名（同一个对象），`copied = demo.copy()` 才是新建一个列表。

````python
demo = [10, 20, 30]
alias = demo            # 直接赋值：两个名字指向同一个列表对象
copied = demo.copy()    # 浅拷贝：新的列表对象，元素逐个复制

alias[0] = 999          # 通过 alias 修改列表内容

print("demo  :", demo)
print("alias :", alias)
print("copied:", copied)
print()
print("alias is demo  ->", alias is demo, "（同一对象，改 alias 就是改 demo）")
print("copied is demo ->", copied is demo, "（不同对象，互不影响）")
print("原始成绩未被影响:", scores)
````

**结论：**`alias = demo` 只是增加了同一个对象的引用，通过 `alias` 修改元素会同时改变
`demo`；`demo.copy()` 返回新列表，元素是逐个复制的，修改副本不影响原列表。
因此本任务使用单独的 `demo` 列表做演示，`scores` 始终保持原始数据。

## 6. 元组的不可变性与 `zip()` 的长度截断

````python
record = ("Alice", 85, True)     # 用元组保存一条学生记录
print("记录:", record, "| 长度:", len(record))

try:
    record[1] = 90               # 元组不可变，元素赋值失败
except TypeError as e:
    print("捕获异常 -> TypeError:", e)

name, score, passed = record     # 但可以解包读取
print("解包结果:", name, score, passed)

# 长度不同的两个序列使用 zip()
short = ["a", "b"]
long = [1, 2, 3, 4]
print("\nzip(短, 长):", list(zip(short, long)))
print("zip(长, 短):", list(zip(long, short)))
print("长度相同    :", list(zip(["a", "b", "c"], [1, 2, 3])))
````

**结论：**

1. 元组创建后**不能修改元素**，`record[1] = 90` 抛出
   `TypeError: 'tuple' object does not support item assignment`；但元组可以整体解包。
2. `zip()` 同时遍历多个序列，遇到**最短的序列结束时就停止**。上面两组 `zip` 都只得到 2 个
   元组，超出的元素被忽略；若需要补齐较短序列，应使用 `itertools.zip_longest()`。

## 3. `parse_student(line)` 函数

把成绩转成整数，并检查字段数量、姓名是否为空、成绩是否在 0~100 范围内；无效记录抛 `ValueError`。

````python
def parse_student(line):
    """把一行 "name,score" 解析为字典；无效记录抛出 ValueError。"""
    fields = line.split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(fields)}")

    name, score_text = fields[0].strip(), fields[1].strip()
    if not name:
        raise ValueError("姓名不能为空")

    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是整数: {score_text!r}")

    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围: {score}")

    return {"name": name, "score": score}


print("有效记录:", parse_student("Alice,85"))
````

## 4. 逐行处理文件，捕获无效记录并继续

记录行号与错误原因，有效记录保存到字典组成的列表。

````python
valid_records = []
invalid_records = []

with open(STUDENTS_TXT, "r", encoding="utf-8") as f:
    next(f)                                  # 跳过表头
    for lineno, raw_line in enumerate(f, start=2):
        line = raw_line.strip()
        if not line:
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as e:
            invalid_records.append({"line_no": lineno, "line": line, "reason": str(e)})

print(f"有效记录 {len(valid_records)} 条，无效记录 {len(invalid_records)} 条")
print("\n无效记录明细：")
for item in invalid_records:
    print(f"  第 {item['line_no']} 行 {item['line']!r} -> {item['reason']}")

print("\n有效记录：")
for record in valid_records:
    print(" ", record)
````

## 5. 写入 `report.txt`，并演示 `w` 与 `a` 模式

统计有效记录数、无效记录数、平均分和及格率。

````python
valid_count = len(valid_records)
invalid_count = len(invalid_records)
average = sum(r["score"] for r in valid_records) / valid_count
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count

REPORT_TEXT = (
    "实验01 数据文件统计报告\n"
    f"有效记录数: {valid_count}\n"
    f"无效记录数: {invalid_count}\n"
    f"平均分    : {average:.2f}\n"
    f"及格率    : {pass_rate:.1%}\n"
)

with open(REPORT_TXT, "w", encoding="utf-8") as f:
    f.write(REPORT_TEXT)

print("— 用 w 模式写入正式统计内容后读取：")
with open(REPORT_TXT, encoding="utf-8") as f:
    print(f.read(), end="")
````

````python
# a 模式：在文件末尾追加，原有内容保留
with open(REPORT_TXT, "a", encoding="utf-8") as f:
    f.write("追加说明: 本次统计只包含成绩合法的记录（a 模式追加，原内容保留）\n")

print("— a 模式追加后：")
with open(REPORT_TXT, encoding="utf-8") as f:
    print(f.read(), end="")

# w 模式：清空原有内容后重写
with open(REPORT_TXT, "w", encoding="utf-8") as f:
    f.write(REPORT_TEXT)

print("— 再用 w 模式重写后（之前的追加内容被清空）：")
with open(REPORT_TXT, encoding="utf-8") as f:
    print(f.read(), end="")
````

**`w` 与 `a` 模式的区别：**

| 模式 | 对已有内容的处理 | 写入位置 |
| --- | --- | --- |
| `w` | **清空**（截断为 0 字节）后写入 | 从文件开头开始 |
| `a` | **保留**原有内容 | 从文件末尾追加 |

两种模式在文件不存在时都会自动创建。所以“覆盖式报告”用 `w`，而“累积日志”应使用 `a`。

# 实验5：类、继承与对象复制

## 1. `Student` 类

````python
class Student:
    """学生：姓名与成绩。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩不低于 60 分即为及格。"""
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, {'及格' if self.is_passed() else '不及格'})"
````

## 2. 为实验 4 的每条有效记录创建 `Student` 对象

````python
students = [Student(record["name"], record["score"]) for record in valid_records]

for student in students:
    print(student, "| is_passed():", student.is_passed())

print("\n及格人数:", sum(1 for s in students if s.is_passed()), "/", len(students))
````

## 3. `GraduateStudent` 继承自 `Student`

增加 `research_topic` 属性，用 `super().__init__()` 初始化父类属性，并重写 `__str__()`。

````python
class GraduateStudent(Student):
    """研究生：在学生基础上增加研究方向。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)          # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic}, {status})")


graduate = GraduateStudent("Alice", 85, "机器学习")
print(graduate)
print("is_passed()（继承自父类）:", graduate.is_passed())
print("isinstance(graduate, Student):", isinstance(graduate, Student))
print("MRO:", [cls.__name__ for cls in GraduateStudent.__mro__])
````

## 4. 直接赋值、`copy.copy()` 与 `copy.deepcopy()`

对包含嵌套列表的对象 `{"name": "Alice", "scores": [85, 90]}`，分别用三种方式复制，
修改内部的列表，比较结果。每次比较都从**新建的原始对象**开始。

````python
import copy


def fresh_student():
    """每次调用都新建一个原始对象，保证三次比较互不干扰。"""
    return {"name": "Alice", "scores": [85, 90]}


print("原始结构模板:", fresh_student())

# 1) 直接赋值
origin1 = fresh_student()
assigned = origin1
assigned["scores"].append(100)
print("\n[直接赋值] assigned = origin1; 修改 assigned 的内层列表")
print("  origin1 :", origin1)
print("  assigned:", assigned)
print("  origin1 is assigned ->", origin1 is assigned)

# 2) 浅拷贝
origin2 = fresh_student()
shallow = copy.copy(origin2)
shallow["scores"].append(100)
print("\n[copy.copy 浅拷贝] 只复制最外层容器")
print("  origin2 :", origin2)
print("  shallow :", shallow)
print("  内层列表是同一对象 ->", shallow["scores"] is origin2["scores"])

# 3) 深拷贝
origin3 = fresh_student()
deep = copy.deepcopy(origin3)
deep["scores"].append(100)
print("\n[copy.deepcopy 深拷贝] 递归复制所有层级")
print("  origin3 :", origin3)
print("  deep    :", deep)
print("  内层列表是同一对象 ->", deep["scores"] is origin3["scores"])
````

````python
# 补充：浅拷贝对“顶层键”的修改不会影响原对象
origin4 = fresh_student()
shallow2 = copy.copy(origin4)
shallow2["name"] = "Bob"          # 只是替换顶层键的指向
print("浅拷贝改顶层键 -> 原对象:", origin4, "| 副本:", shallow2)
````

**结论：**

| 方式 | 是否新建对象 | 嵌套的可变对象 | 修改内层列表是否影响原对象 |
| --- | --- | --- | --- |
| `assigned = origin` | 否（同一个对象） | 共享 | 会 |
| `copy.copy(origin)` | 是（新的外壳） | **共享** | 会 |
| `copy.deepcopy(origin)` | 是 | 递归复制，完全独立 | 不会 |

浅拷贝只复制最外层容器：替换或修改顶层键不会影响原对象，但内层列表仍与原对象共享，
所以 `shallow["scores"].append(100)` 会“顺带”改到原对象。深拷贝递归复制所有层级，
两者彻底独立。对嵌套结构做独立修改时必须使用 `copy.deepcopy()`。

## 实验总结

1. **运算与类型**：`/` 总返回浮点数，`//`、`%`、`**` 在整数运算中保持 `int`；`int()` 对浮点数
   是向零截断；非法字符串转换抛 `ValueError`。
2. **字符串不可变**：所有“修改”方法都返回新对象，逐个字符替换需用切片拼接或 `replace()`。
3. **函数与循环**：`if/elif` 顺序判断区间；`for`/`while` 求和解一致；递归与迭代阶乘结果一致；
   `*args` 把位置参数收集成元组；元组解包可直接接收多返回值。
4. **容器与统计**：`zip` 建字典、`items` 遍历、`enumerate` 编号、推导式筛选与变换；
   赋值是引用同一对象，`copy()` 才是新列表；元组不可变；`zip` 按最短序列截断。
5. **文件与异常**：`with open(..., encoding="utf-8")` 自动关闭文件；`w` 清空重写、`a` 追加保留；
   `parse_student` 用 `ValueError` 表达无效记录，逐行处理时捕获异常并继续；
   `pickle` 用于对象序列化，`FileNotFoundError` 用于缺失文件。
6. **类与对象**：`__init__` 初始化属性、`__str__` 控制打印；`GraduateStudent` 通过
   `super().__init__()` 复用父类初始化并重写 `__str__`；嵌套结构中只有 `deepcopy` 能完全独立。
