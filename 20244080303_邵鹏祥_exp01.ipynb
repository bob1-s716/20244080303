```python
# 实验2-1 成绩等级
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("成绩必须在0到100之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in [95, 90, 85, 80, 70, 60, 59, 0]:
    print(score, score_level(score))

try:
    score_level(101)
except ValueError as e:
    print("ValueError:", e)
```

```python
# 实验2-3 阶乘，递归和迭代
def fact_rec(n):
    if n == 0:
        return 1
    return n * fact_rec(n - 1)


def fact_iter(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result


for n in [0, 1, 5, 10]:
    print(n, fact_rec(n), fact_iter(n))
```

```python
# 实验4-1 工作目录与数据目录
print(os.getcwd())

data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(data_dir)
```

```python
# 实验4-2 写入并读取数据文件
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

path = os.path.join(data_dir, "students.txt")
with open(path, "w", encoding="utf-8") as f:
    f.write(content)

with open(path, encoding="utf-8") as f:
    lines = f.read().splitlines()

print(lines[0])
for line in lines[1:]:
    print(line.split(","))
```

```python
# 实验4-3 解析一行记录
def parse_student(line):
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("字段数不是2")
    name = parts[0].strip()
    if name == "":
        raise ValueError("姓名为空")
    try:
        score = int(parts[1])
    except ValueError:
        raise ValueError("成绩不是整数")
    if score < 0 or score > 100:
        raise ValueError("成绩不在0到100之间")
    return {"name": name, "score": score}


print(parse_student("Alice,85"))
```

```python
# 实验4-4 逐行处理，跳过无效记录
students = []
errors = []

with open(path, encoding="utf-8") as f:
    next(f)
    for no, line in enumerate(f, 2):
        line = line.strip()
        if line == "":
            continue
        try:
            students.append(parse_student(line))
        except ValueError as e:
            errors.append((no, line, str(e)))

print(len(students), len(errors))
for item in errors:
    print(item)
print(students)
```

```python
# 实验4-5 写统计报告，w模式会清空，a模式在末尾追加
report = os.path.join(data_dir, "report.txt")
valid_scores = [student["score"] for student in students]
avg_score = sum(valid_scores) / len(valid_scores)
rate = len([s for s in valid_scores if s >= 60]) / len(valid_scores)

report_text = ("有效记录数: %d\n无效记录数: %d\n平均分: %.2f\n及格率: %.1f%%\n"
               % (len(students), len(errors), avg_score, rate * 100))

with open(report, "w", encoding="utf-8") as f:
    f.write(report_text)

with open(report, encoding="utf-8") as f:
    print(f.read())

with open(report, "a", encoding="utf-8") as f:
    f.write("追加一行测试\n")

with open(report, encoding="utf-8") as f:
    print(f.read())

with open(report, "w", encoding="utf-8") as f:
    f.write(report_text)

with open(report, encoding="utf-8") as f:
    print(f.read())
```

```python
# 实验5-1 Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return "%s, %d, %s" % (self.name, self.score, "及格" if self.is_passed() else "不及格")
```

```python
# 实验5-2 为有效记录创建对象
objs = [Student(student["name"], student["score"]) for student in students]
for obj in objs:
    print(obj, obj.is_passed())
```

```python
# 实验5-3 GraduateStudent继承Student
class GraduateStudent(Student):
    def __init__(self, name, score, topic):
        super().__init__(name, score)
        self.topic = topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return "%s, %d, %s, 研究方向: %s" % (self.name, self.score, status, self.topic)


g = GraduateStudent("Alice", 85, "机器学习")
print(g)
print(g.is_passed())
```
