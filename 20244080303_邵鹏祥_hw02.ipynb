```python
# bonjour

def is_palindrome(s):
    chars = [c.lower() for c in s if c.isalnum()]
    return chars == chars[::-1]


print(is_palindrome("level"))
print(is_palindrome("tacocat"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("rats live on no evil star"))
print(is_palindrome("Was it a car or a cat I saw?"))
print(is_palindrome("hello"))
```

```python
def is_abecedarian(s):
    letters = [c.lower() for c in s if c != " "]
    for i in range(len(letters) - 1):
        if letters[i] > letters[i + 1]:
            return False
    return True


print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("dog"))
print(is_abecedarian("cat"))
print(is_abecedarian("abcd efgh xyz"))
```

```python
def remove_vowels(s):
    return "".join(c for c in s if c.lower() not in "aeiou")


print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
print(remove_vowels("Python"))
```

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]


print(list_reverse([1, 2, 3]))
print(list_reverse(["a", "b", "c", "d"]))

try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print("TypeError:", e)
```

```python
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True


print(is_sorted([1, 2, 3, 4]))
print(is_sorted([3, 1, 2]))
print(is_sorted((1, 2, 2, 3)))
print(is_sorted("abc"))
print(is_sorted("acb"))
```

```python
def is_num(x):
    return isinstance(x, (int, float)) and not isinstance(x, bool)


def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("t 必须是元组")
    if not is_num(s):
        raise TypeError("s 必须是整数或浮点数")
    result = []
    for x in t:
        if not is_num(x):
            raise TypeError("向量的元素必须是整数或浮点数")
        result.append(x * s)
    return tuple(result)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((0.5, 1.5), 2))
print(vec_scalar_mult((1, 2, 3), 0.5))

try:
    vec_scalar_mult([1, 2, 3], 2)
except TypeError as e:
    print("TypeError:", e)
```

```python
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量的长度必须相同")
    total = 0.0
    for i in range(len(x)):
        if not is_num(x[i]) or not is_num(y[i]):
            raise TypeError("向量的元素必须是整数或浮点数")
        total += x[i] * y[i]
    return float(total)


print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((1.5, 2), (2, 4)))

try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
```

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple) or len(mx) == 0:
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    n = len(mx[0])
    for row in mx:
        if len(row) != n:
            return False
        for x in row:
            if not is_num(x):
                return False
    return True


print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2, 3), (4, 5, 6))))
print(check_valid_mx(((1, 2), (3,))))
print(check_valid_mx(((1, 2), ("a", 3))))
print(check_valid_mx([[1, 2], [3, 4]]))
print(check_valid_mx(()))
```

```python
def mx_vec_mult(mx, vec):
    if not isinstance(mx, tuple) or not isinstance(vec, tuple):
        raise TypeError("矩阵和向量都必须是元组")
    if not check_valid_mx(mx):
        raise ValueError("不是有效的矩阵")
    if len(vec) != len(mx[0]):
        raise ValueError("向量长度必须等于矩阵的列数")
    result = []
    for row in mx:
        total = 0.0
        for i in range(len(vec)):
            if not is_num(vec[i]):
                raise TypeError("向量的元素必须是整数或浮点数")
            total += row[i] * vec[i]
        result.append(total)
    return tuple(result)


print(mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
print(mx_vec_mult(((1, 2, 3), (4, 5, 6)), (1, 0, 1)))

try:
    mx_vec_mult(((1, 2), (3, 4)), (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
```

```python
class Vector:
    # 维度 dim 和坐标 elems 用实例属性：每个向量各有各的值和长度，
    # 类属性会被所有实例共享，改一个就影响全部，只适合放常量。
    def __init__(self, dim, elems=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elems is None:
            elems = [0] * dim
        if not isinstance(elems, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elems) != dim:
            raise ValueError("维度与元素个数不一致")
        for x in elems:
            if not is_num(x):
                raise TypeError("元素必须是整数或浮点数")
        self.dim = dim
        self.elems = tuple(elems)

    def __repr__(self):
        return "Vector(%d, %r)" % (self.dim, self.elems)

    def check(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")
        return self.elems == other.elems

    def __lt__(self, other):
        self.check(other)
        return self.elems < other.elems

    def __le__(self, other):
        self.check(other)
        return self.elems <= other.elems

    def __gt__(self, other):
        self.check(other)
        return self.elems > other.elems

    def __ge__(self, other):
        self.check(other)
        return self.elems >= other.elems

    def dot(self, other):
        self.check(other)
        total = 0.0
        for i in range(self.dim):
            total += self.elems[i] * other.elems[i]
        return float(total)

    def __mul__(self, other):
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度必须相同")
            return Vector(self.dim, tuple(a * b for a, b in zip(self.elems, other.elems)))
        if is_num(other):
            return Vector(self.dim, tuple(a * other for a in self.elems))
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)
```

```python
v = Vector(2, (2, 4))
w = Vector(2, (2, 5))

print(v == w)
print(v < w)
print(v <= w)
print(w > v)
print(w >= v)
print(Vector(3, (1, 2, 3)) == Vector(3, (1, 2, 3)))

try:
    print(v < Vector(3))
except ValueError as e:
    print("ValueError:", e)
```

```python
print(v.dot(w))
print(v.dot(Vector(2, (1, 1))))

try:
    v.dot(5)
except TypeError as e:
    print("TypeError:", e)

try:
    v.dot(Vector(4))
except ValueError as e:
    print("ValueError:", e)
```

```python
print(2 * v)
print(v * 2)
print(v * w)
print(Vector(2, (2, 3)) * Vector(2, (4, 5)))
print(v * 1.5)

try:
    v * Vector(3)
except ValueError as e:
    print("ValueError:", e)
```
